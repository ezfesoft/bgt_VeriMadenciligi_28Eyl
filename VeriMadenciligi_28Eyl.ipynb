```python
import pandas as pd

df = pd.read_csv('hafta1_sosyal_medya.csv')
print(df.shape) # Satır ve sütun sayısı
df.head(3)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df.info()
```

```python
df.describe()
```

```python
ozellikler = df.columns[:3]
print('İncelenen İlk 3 Özellik:', list(ozellikler))
```

```python
filtre = df[df.columns[2]] > df[df.columns[2]].mean()
secilen_veriler = df[filtre]
print('Ortalama Üstü Kayıt Sayısı:', len(secilen_veriler))
```

```python
platform_ortalama = df.groupby('Platform')['Takipci_Sayisi'].mean()

plt.figure(figsize=(8, 4))
plt.bar(platform_ortalama.index, platform_ortalama.values, color=['blue', 'orange', 'green', 'red'])

plt.title('Platformlara Göre Ortalama Takipçi Sayısı')
plt.xlabel('Sosyal Medya Platformu')
plt.ylabel('Ortalama Takipçi Sayısı')

plt.show()
```
